# 01 — Загрузка своих файлов в MLflow

Укажите **любой путь** к файлу или папке на вашем компьютере (или в `/home/jovyan/host-data/` если смонтировали `DS_DATA_HOST_PATH`).

Результат — run в MLflow. Откройте http://localhost:5000

```python
import os
import sys
from pathlib import Path

ROOT = Path('/home/jovyan/work') if Path('/home/jovyan/work').exists() else Path('..').resolve()
sys.path.insert(0, str(ROOT))

os.environ.setdefault('MLFLOW_TRACKING_URI', 'http://mlflow:5000')
os.environ.setdefault('MLFLOW_S3_ENDPOINT_URL', 'http://minio:9000')
os.environ.setdefault('AWS_ACCESS_KEY_ID', 'minio')
os.environ.setdefault('AWS_SECRET_ACCESS_KEY', 'changeme')

# ← ваши пути (Windows в Docker Jupyter: /home/jovyan/host-data/...)
LOCAL_PATH = Path('/home/jovyan/host-data/my_data.csv')  # или r'C:\Users\YOU\file.csv' локально
EXPERIMENT = 'my-experiment'
RUN_NAME = 'upload-1'
YOUR_LOGIN = 'ваш_логин'  # логин FORTRESS / Keycloak

print('Путь:', LOCAL_PATH, '— exists:', LOCAL_PATH.exists())
```

```python
from fortress.ds_mlflow_upload import log_local_files

run_id, uri = log_local_files(
    [LOCAL_PATH],
    experiment=EXPERIMENT,
    run_name=RUN_NAME,
    owner=YOUR_LOGIN,
)
print('run_id:', run_id)
print('artifacts:', uri)
```

Готово — файлы в MLflow. Откройте http://localhost:5000 → эксперимент `my-experiment`.

**DATA gate (опционально):** если нужен pipeline FORTRESS, используйте `scripts/ds_upload_dataset.py upload ...` или вкладку **Данные** → «Регистрация для pipeline».
